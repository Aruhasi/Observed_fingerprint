```python
import src.plot_func as func_plot

# define an asymmetric colormap
from matplotlib.colors import LinearSegmentedColormap, Normalize
from matplotlib.colors import BoundaryNorm
import cartopy.util as cutil
import seaborn as sns
import matplotlib.colors as mcolors
import palettable
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import cartopy.crs as ccrs
import cartopy.util as cutil
from collections import OrderedDict

# === Define model display order ===
model_names = ["CanESM5", "CESM2","IPSL-CM6A-LR", "EC_Earth3", "ACCESS-ESM1-5", "MPI-ESM1-2-LR", "MIROC6", "Multi-model mean"]

# === Ordered slope dictionaries ===
Slope_dict = OrderedDict({
    "CanESM5": CanESM_beta_mean,
    "CESM2": CESM2_beta_mean,
    "IPSL-CM6A-LR": IPSL_beta_mean,
    "EC_Earth3": EC_Earth_beta_mean,
    "ACCESS-ESM1-5": ACCESS_beta_mean,
    "MPI-ESM1-2-LR": MPI_ESM_beta_mean,
    "MIROC6": MIROC_beta_mean,
    "Multi-model mean": MMEM_beta_mean,
})

Slope_dict_wrt_MMLE = OrderedDict({
    "CanESM5": CanESM_MMLE_beta_mean,
    "CESM2": CESM2_MMLE_beta_mean,
    "IPSL-CM6A-LR": IPSL_MMLE_beta_mean,
    "EC_Earth3": EC_Earth_MMLE_beta_mean,
    "ACCESS-ESM1-5": ACCESS_MMLE_beta_mean,
    "MPI-ESM1-2-LR": MPI_ESM_MMLE_beta_mean,
    "MIROC6": MIROC_MMLE_beta_mean,
    "Multi-model mean": MMEM_MMLE_beta_mean,
})

# === Plotting settings ===
levels = np.arange(-1.0, 1.1, 0.1)
extend = 'both'
cmap = 'RdBu_r'

fig = plt.figure(figsize=(14, 20))
gs = gridspec.GridSpec(8, 2, wspace=0.01, hspace=0.15)

# === Loop over models ===
for i, model in enumerate(model_names):
    # ---- Column 1: wrt itself ----
    slope_self = Slope_dict[model]
    ax1 = fig.add_subplot(gs[i, 0], projection=ccrs.Robinson(central_longitude=180))
    
    slope_self_cyclic, lon_cyclic = cutil.add_cyclic_point(slope_self.values, coord=slope_self['lon'].values)
    contour_obj = func_plot.plot_data(
        slope_self_cyclic, slope_self['lat'], lon_cyclic,
        levels=levels, extend=extend, cmap=cmap, title=" ",
        ax=ax1, show_xticks=False, show_yticks=False
    )
    ax1.text(0.5, 1.08, rf"{model} $\beta^*$ (w.r.t. its GSAT)", ha='center', va='center', transform=ax1.transAxes, fontsize=14)

    # ---- Column 2: wrt MMLE ----
    slope_mmem = Slope_dict_wrt_MMLE[model]
    ax2 = fig.add_subplot(gs[i, 1], projection=ccrs.Robinson(central_longitude=180))
    
    slope_mmem_cyclic, _ = cutil.add_cyclic_point(slope_mmem.values, coord=slope_mmem['lon'].values)
    contour_obj_for_cbar = func_plot.plot_data(
        slope_mmem_cyclic, slope_mmem['lat'], lon_cyclic,
        levels=levels, extend=extend, cmap=cmap, title=" ",
        ax=ax2, show_xticks=False, show_yticks=False
    )
    ax2.text(0.5, 1.08, rf"{model} $\beta^*$ (w.r.t. MMLE GSAT)", ha='center', va='center', transform=ax2.transAxes, fontsize=14)

# === Colorbar ===
cbar_ax = fig.add_axes([0.25, 0.08, 0.5, 0.012])
cbar = plt.colorbar(contour_obj_for_cbar, cax=cbar_ax, orientation='horizontal', extend=extend)
cbar.set_label("Regression Coefficient (°C/°C)", fontsize=14)
cbar.ax.tick_params(labelsize=12)

# === Save and Show ===
plt.tight_layout(rect=[0, 0.06, 1, 1])
fig.savefig("MMLE_beta_pattern_comparison.png", dpi=300, bbox_inches='tight')
fig.savefig("MMLE_beta_pattern_comparison.pdf", dpi=300, bbox_inches='tight')
plt.show()
```
